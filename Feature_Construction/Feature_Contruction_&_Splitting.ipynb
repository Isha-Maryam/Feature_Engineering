# Feature Construction

making an extra column by using sibling and parch column to estimate faily size traveling toether or alone.

```python
x['Family_size']= x['SibSp'] + x['Parch'] + 1
```

```python
x.head()
```

```python
def myfunc(num):
  if num==1: #alone
    return 0
  elif num>=1 and num<=4: #small family
    return 1
  else: #large family
    return 2
```

```python
myfunc(4)
```

```python
x['Family_size']=x['Family_size'].apply(myfunc)
```

```python
x.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),x,y,scoring='accuracy',cv=20))
```
